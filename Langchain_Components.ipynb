# Langchain Components


LangChain is an open-source framework that makes it easier to build applications powered by large language models (LLMs)

The 6 Main Components of LangChain: 

1. Model = does the actual language processing

2. Prompt = tells it what to do

3. Chain = connects steps together

4. Memory = remembers past context

5. Indexes = fetches external knowledge

6. Agent = decides what to do next

## 1. Models

In LangChain, models are the core AI engines that perform the main tasks of your application — from generating text to retrieving relevant information.

LangChain doesn’t create its own models; instead, it connects to existing LLMs, chat models, and embedding models from providers like OpenAI, Anthropic, Google, HuggingFace, Cohere, Ollama, etc.

### Types of Models in LangChain

#### A.LLMs (Large Language Models)
- Definition: Take plain text as input and return plain text as output.
- Best for: Summarization, text generation, Q&A, translations.

- Examples:
- gpt-4, gpt-3.5-turbo-instruct (OpenAI)
- claude-3-opus (Anthropic)
- llama-3-70b (Meta, via HuggingFace or Ollama)

```python
from langchain_openai import OpenAI
from dotenv import load_dotenv

# load the api from .env file
load_dotenv()

llm = OpenAI(model="gpt-3.5-turbo-instruct")

result = llm.invoke("What is the largest district of Banglades")

print(result)
```

#### B. Chat Models
Definition: Similar to LLMs, but optimized for multi-turn conversation format with roles like:

- System (sets behavior)

- User (asks questions)

- Assistant (model’s replies)

Best for: Chatbots, customer support, assistants.

Examples:

- gpt-4o, gpt-3.5-turbo (OpenAI)

- claude-3-sonnet (Anthropic)

- gemini-pro (Google)

```python
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

# Initialize the chat model
chat_model = ChatOpenAI(model_name="gpt-4", temperature=0.5)

# Create a conversation
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Tell me 3 interesting facts about the moon.")
]

# Send messages and get a reply
response = chat_model.invoke(messages)
print(response.content)
```

#### C. Embedding Models
Definition: Convert text into numerical vectors that capture meaning.

Best for: Search, recommendation systems, semantic similarity, Retrieval-Augmented Generation (RAG).

Examples:

 - text-embedding-ada-002 (OpenAI)

 - all-MiniLM-L6-v2 (HuggingFace)

 - bge-base-en (BAAI)

```python
from langchain_openai import OpenAIEmbeddings

# Initialize the embedding model
embedding_model = OpenAIEmbeddings(model="text-embedding-ada-002")

# Convert text into a vector
vector = embedding_model.embed_query("Artificial Intelligence is amazing")
print(f"Vector length: {len(vector)}")
print(f"First 5 numbers: {vector[:5]}")
```
