# Langchain Components


LangChain is an open-source framework that makes it easier to build applications powered by large language models (LLMs)

The 6 Main Components of LangChain: 

1. Model = does the actual language processing

2. Prompt = tells it what to do

3. Chain = connects steps together

4. Memory = remembers past context

5. Indexes = fetches external knowledge

6. Agent = decides what to do next

## 1. Models

In LangChain, models are the core AI engines that perform the main tasks of your application — from generating text to retrieving relevant information.

LangChain doesn’t create its own models; instead, it connects to existing LLMs, chat models, and embedding models from providers like OpenAI, Anthropic, Google, HuggingFace, Cohere, Ollama, etc.

### Types of Models in LangChain

#### A.LLMs (Large Language Models)
- Definition: Take plain text as input and return plain text as output.
- Best for: Summarization, text generation, Q&A, translations.

- Examples:
- gpt-4, gpt-3.5-turbo-instruct (OpenAI)
- claude-3-opus (Anthropic)
- llama-3-70b (Meta, via HuggingFace or Ollama)

```python
from langchain_openai import OpenAI
from dotenv import load_dotenv

# load the api from .env file
load_dotenv()

llm = OpenAI(model="gpt-3.5-turbo-instruct")

result = llm.invoke("What is the largest district of Banglades")

print(result)
```

#### B. Chat Models
Definition: Similar to LLMs, but optimized for multi-turn conversation format with roles like:

- System (sets behavior)

- User (asks questions)

- Assistant (model’s replies)

Best for: Chatbots, customer support, assistants.

Examples:

- gpt-4o, gpt-3.5-turbo (OpenAI)

- claude-3-sonnet (Anthropic)

- gemini-pro (Google)

```python
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

# Initialize the chat model
chat_model = ChatOpenAI(model_name="gpt-4", temperature=0.5)

# Create a conversation
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Tell me 3 interesting facts about the moon.")
]

# Send messages and get a reply
response = chat_model.invoke(messages)
print(response.content)
```

#### C. Embedding Models
Definition: Convert text into numerical vectors that capture meaning.

Best for: Search, recommendation systems, semantic similarity, Retrieval-Augmented Generation (RAG).

Examples:

 - text-embedding-ada-002 (OpenAI)

 - all-MiniLM-L6-v2 (HuggingFace)

 - bge-base-en (BAAI)

```python
from langchain_openai import OpenAIEmbeddings

# Initialize the embedding model
embedding_model = OpenAIEmbeddings(model="text-embedding-ada-002")

# Convert text into a vector
vector = embedding_model.embed_query("Artificial Intelligence is amazing")
print(f"Vector length: {len(vector)}")
print(f"First 5 numbers: {vector[:5]}")
```

### 2. Prompts in LangChain
What Are Prompts?
- In LangChain, a prompt is the structured text (or message) you send to the model to tell it exactly what to do.

- Without a good prompt, the model won’t produce the output you want.

- LangChain provides PromptTemplate and ChatPromptTemplate to make prompts dynamic, reusable, and role-aware.

- Think of a prompt as the instructions given to the AI “brain.”

#### A. Dynamic & Reusable Prompts
- Definition: A prompt where certain parts are variables, so you can reuse the same structure for multiple inputs.

- Why Useful: Avoids writing a new prompt for every input; makes it easy to scale.

- LangChain Tool: PromptTemplate

```python
from langchain.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

# Define a reusable prompt with a variable
prompt = PromptTemplate(
    input_variables=["topic"],
    template="Write a short blog post about {topic}."
)

# Fill in the variable dynamically
filled_prompt = prompt.format(topic="Artificial Intelligence")
print("Generated Prompt:", filled_prompt)

# Send to LLM
llm = ChatOpenAI()
response = llm.invoke(filled_prompt)
print(response)
```

How it works:

- {topic} is replaced dynamically each time.

- You can reuse this template for hundreds of topics.

#### B. Role-Based Prompts
- Definition: Prompts structured with roles (System, User, Assistant) to give context to the conversation.

- Why Useful: Helps control how the model behaves.

- LangChain Tool: ChatPromptTemplate

```python
from langchain.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

# Define a chat prompt with roles
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a professional fitness coach."),
    ("user", "Suggest a {duration}-minute workout for beginners.")
])

# Fill in variable
filled_prompt = chat_prompt.format_messages(duration=30)

# Send to Chat Model
chat_model = ChatOpenAI()
response = chat_model.invoke(filled_prompt)
print(response.content)
```

How it works:

- System role defines the assistant’s behavior.

- User role gives the query.

- You can add Assistant role examples for better context.

#### C. Few-Shot Prompting
- Definition: Providing a few examples in the prompt to guide the model’s response style and format.

- Why Useful: Improves accuracy by showing the model how to answer.

- LangChain Tool: Can be done with PromptTemplate or ChatPromptTemplate.

```python
from langchain.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

# Few-shot example template
template = """
Classify the sentiment of the text as Positive, Negative, or Neutral.

Example 1:
Text: "I love this product, it’s amazing!"
Sentiment: Positive

Example 2:
Text: "This is the worst service I have ever experienced."
Sentiment: Negative

Example 3:
Text: "{text}"
Sentiment:
"""

prompt = PromptTemplate(input_variables=["text"], template=template)

llm = ChatOpenAI()
response = llm.invoke(prompt.format(text="The movie was okay, nothing special."))
print(response)
```

How it works:

- The first two examples “teach” the model how to respond.

- The last example is the actual query.

### 3. Chains
Definition:
- In LangChain, Chains connect multiple components (Models, Prompts, Tools, etc.) in a sequence or structure so that the output of one step becomes the input for the next.
This helps you create multi-step workflows without writing complex glue code manually.

Why Chains Are Needed
- Instead of a single LLM call, you can build multi-step pipelines.

- Makes it easier to reuse, debug, and extend your LLM logic.

- Allows parallel, conditional, and sequential flows.

#### A. Parallel Chains
- Run multiple chains at the same time and combine their results.

- Example: Summarize text in English and French simultaneously.

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.schema.runnable import RunnableParallel
from dotenv import load_dotenv
load_dotenv()

model1 = ChatGroq(model="llama-3.1-8b-instant", temperature=0.0, max_retries=2)

model2 = ChatGroq(model="qwen/qwen3-32b")

prompt1 = PromptTemplate(
    template='Generate short and simple notes from the following text \n {text}',
    input_variables=['text']
)

prompt2 = PromptTemplate(
    template='Generate 5 short question answers from the following text \n {text}',
    input_variables=['text']
)

prompt3 = PromptTemplate(
    template='Merge the provided notes and quiz into a single document \n notes -> {notes} and quiz -> {quiz}',
    input_variables=['notes', 'quiz']
)

parser = StrOutputParser()

parallel_chain = RunnableParallel({
    'notes':prompt1 | model1 | parser ,
    'quiz':prompt2 | model2 | parser
})

merge_chain = prompt3 | model1 | parser

chain = parallel_chain | merge_chain

text = """
Support vector machines (SVMs) are a set of supervised learning methods used for classification, regression and outliers detection.

The advantages of support vector machines are:

Effective in high dimensional spaces.

Still effective in cases where number of dimensions is greater than the number of samples.

Uses a subset of training points in the decision function (called support vectors), so it is also memory efficient.

Versatile: different Kernel functions can be specified for the decision function. Common kernels are provided, but it is also possible to specify custom kernels.

The disadvantages of support vector machines include:

If the number of features is much greater than the number of samples, avoid over-fitting in choosing Kernel functions and regularization term is crucial.

SVMs do not directly provide probability estimates, these are calculated using an expensive five-fold cross-validation (see Scores and probabilities, below).

The support vector machines in scikit-learn support both dense (numpy.ndarray and convertible to that by numpy.asarray) and sparse (any scipy.sparse) sample vectors as input. However, to use an SVM to make predictions for sparse data, it must have been fit on such data. For optimal performance, use C-ordered numpy.ndarray (dense) or scipy.sparse.csr_matrix (sparse) with dtype=float64.
"""

result = chain.invoke({'text':text})

print(result)

chain.get_graph().print_ascii()
```

#### B. Conditional Chains
Run different chains depending on conditions (e.g., language detection).

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser, PydanticOutputParser
from langchain.schema.runnable import RunnableParallel, RunnableBranch, RunnableLambda
from pydantic import BaseModel, Field
from typing import Literal
from dotenv import load_dotenv
load_dotenv()

model = ChatGroq(model="llama-3.1-8b-instant", temperature=0.0, max_retries=2)

parser1 = StrOutputParser()

class Feedback(BaseModel):
    sentiment:Literal['positive','negative'] = Field(description="Give the sentiment of the feedback")

parser2 = PydanticOutputParser(pydantic_object=Feedback)


prompt1 = PromptTemplate(
    template='Classify the sentiment of the following feedback text into postive or negative \n {feedback} \n {format_instruction}',
    input_variables=['feedback'],
    partial_variables={'format_instruction':parser2.get_format_instructions()}
)

classifier_chain = prompt1 | model | parser2

prompt2 = PromptTemplate(
    template='Write an appropriate response to this positive feedback \n {feedback}',
    input_variables=['feedback']
)

prompt3 = PromptTemplate(
    template='Write an appropriate response to this negative feedback \n {feedback}',
    input_variables=['feedback']
)

branch_chain = RunnableBranch(
    (lambda x:x.sentiment == 'positive', prompt2 | model | parser1),
    (lambda x:x.sentiment == 'negative', prompt3 | model | parser1),
    RunnableLambda(lambda x: "could not find sentiment")
)

chain = classifier_chain | branch_chain

print(chain.invoke({'feedback': 'This is a beautiful phone'}))

chain.get_graph().print_ascii()

```

#### C. Sequential Chains
The output of one chain is the input of the next.

Example:

- Generate a blog outline.

- Write the full blog post.

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()

model = ChatGroq(model="llama-3.1-8b-instant", temperature=0.0, max_retries=2)

template1 = PromptTemplate(
    template="Generate a detailed report based on the {topic}",
    input_variables=['topic']
)

template2 = PromptTemplate(
    template='Generate a 5 pointer summary from the following text \n {text}',
    input_variables=['text']
)

parser = StrOutputParser()

chain = template1 | model | parser | template2 | model | parser

result = chain.invoke({'topic':'Education System of Bangladesh'})

print(result)

chain.get_graph().print_ascii()
```

#### 4. Load Data from a Directory

```python
from langchain.document_loaders import DirectoryLoader
from langchain.document_loaders import PyPDFLoader

# Load all PDFs from folder
loader = DirectoryLoader("docs_folder", glob="*.pdf", loader_cls=PyPDFLoader)
documents = loader.load()
print(f"Loaded {len(documents)} documents.")
```

1. DirectoryLoader("docs_folder", glob="*.pdf", loader_cls=PyPDFLoader)
DirectoryLoader:
This is a LangChain utility class that loads multiple documents from a directory (folder).

"docs_folder":
This is the path to the folder where your documents are stored. The loader will look inside this folder for files.

glob="*.pdf":
The glob argument specifies a pattern to match files in the folder.

"*.pdf" means only load files that end with .pdf (all PDF files).

You could change this to "*.txt", "*.md", or any other file extension to load different types.

loader_cls=PyPDFLoader:
This tells the DirectoryLoader which document loader to use for each file it finds.

Here, PyPDFLoader is a loader specialized for loading PDF files (it reads PDFs and extracts their text).

This is important because DirectoryLoader just scans files; it needs to know how to parse each file type, so you pass the right loader class.

2. documents = loader.load()
This calls the .load() method on the DirectoryLoader.

It loads all matching files in the folder using the specified loader (PyPDFLoader for PDFs).

The result is a list of Document objects (LangChain's internal format) representing the content of each PDF file.

3. print(f"Loaded {len(documents)} documents.")
This prints out the total number of documents loaded from the folder.

len(documents) counts how many documents were extracted.
